```python
#just an example here u can use ur data to see outliers
import numpy as np

data = np.array([12,13,14,15,16,100])
Q1, Q3 = np.percentile(data, [25, 75])
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = data[(data < lower) | (data > upper)]
print("Outliers:", outliers)
```
